```python
import pandas as pd
import numpy as np
from pathlib import Path

data_path = Path("../data/raw/training_setA")

patient_files = list(data_path.glob("*.psv"))

print("Total patients:", len(patient_files))

feature_dataset = []

for file in patient_files:

    patient = pd.read_csv(file, sep="|")

    features = {}

    numeric_columns = patient.select_dtypes(include=np.number).columns
    numeric_columns = numeric_columns.drop("SepsisLabel")

    # Feature engineering
    for column in numeric_columns:

        features[column + "_mean"] = patient[column].mean()
        features[column + "_min"] = patient[column].min()
        features[column + "_max"] = patient[column].max()
        features[column + "_std"] = patient[column].std()
        features[column + "_last"] = patient[column].iloc[-1]

    # Patient label
    features["SepsisLabel"] = patient["SepsisLabel"].max()

    # Save one patient
    feature_dataset.append(features)

# Convert after ALL patients are processed
feature_dataset = pd.DataFrame(feature_dataset)

print(feature_dataset.shape)

feature_dataset.head()
```

```python
from pathlib import Path

output_path = Path("../data/processed")
output_path.mkdir(exist_ok=True)

feature_dataset.to_csv(output_path / "feature_dataset.csv", index=False)

print("Feature dataset saved successfully!")
```
