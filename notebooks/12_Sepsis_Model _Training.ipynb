```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
```

```python
model_a = pd.read_csv("../data/processed/model_a_features.csv")
cohort = pd.read_csv("../data/hf_dataset/sepsis_cohort.csv")
```

```python
print(model_a.shape) # denotes the hourly observation of each patient with particular stay id
print(cohort.shape) #the patient's final medical chart. It tells you the start time and the exact time the event (sepsis) happened, but it doesn't show the hour-by-hour vitals.
model_a.head() # here  the Each row is one hourly observation of one patient during their ICU stay
```

```python
print(cohort.columns.tolist())  # cohort is a subset of dataset and it is a dataset object it consist of data like patient in time and other information related to sepsis like sofa (simply it consist of crucial inforomation sepsis time and icuintime
```

```python
cohort.head()
```

```python
### MERGING COHORT VALUES INTO MODEL_A
model = model_a.merge(
    cohort[
        [
            "stay_id",
            "icu_intime",
            "sepsis_time"
        ]

    ],
    on = "stay_id",
    how = "left"
)
print(model.shape)
model.head()

print(model[["hour",
             "icu_intime",
             "sepsis_time"]].head(10))
```

```python
model["icu_intime"] = pd.to_datetime(model["icu_intime"])
model["sepsis_time"] = pd.to_datetime(model["sepsis_time"])
```

```python
# Current timestamp for each hourly record
model["current_time"] = (
    model["icu_intime"] +
    pd.to_timedelta(model["hour"], unit="h")
)

# Hours remaining until sepsis
model["hours_until_sepsis"] = (
    model["sepsis_time"] - model["current_time"]
).dt.total_seconds() / 3600

model[
    [
        "stay_id",
        "hour",
        "current_time",
        "sepsis_time",
        "hours_until_sepsis"
    ]
].head(15)
```

```python
# 6-hour early warning target

model["target"] = (
    (model["hours_until_sepsis"] >= 0) &
    (model["hours_until_sepsis"] <= 6)
).astype(int)      # astype(int) convert true or false result into 1 or 0 . if patient likely to get sepsis between 1 to 6 hrs then label be 1 other than tht label will be 0

model[
    [
        "hour",
        "hours_until_sepsis",
        "target"
    ]
].head(15) # negative denotes the patient already have sepsis and positive denotes the patient likely to get sepsis for the patient tht doesnt have sepsis will be denotoed by NAT
```

```python
### patient who developed sepsis
septic_patients = model[
    model["sepsis_time"].notna() ### identidies patient who have sepsis notna dropes the NAN in sepsis time which denotes the patient didnt get sepsis
    ]["stay_id"].unique()
### patient who never developed sepsis
non_septic_patients = model[
    model["sepsis_time"].isna()
]["stay_id"].unique()
print("Septic ICU stays:", len(septic_patients))
print("Non-septic ICU stays:", len(non_septic_patients))
```
