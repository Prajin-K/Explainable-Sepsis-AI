```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

data_path = Path("../data/raw/training_setA")
patient_files = list(data_path.glob("*.psv"))
icu_hours = []
for file in patient_files:
    patient = pd.read_csv(file,sep="|")
    icu_hours.append(len(patient))
plt.figure(figsize=(10,5))

plt.hist(icu_hours, bins=50)

plt.title("Distribution of ICU Monitoring Duration")

plt.xlabel("Hours in ICU")
plt.ylabel("Number of Patients")

plt.show()
```
