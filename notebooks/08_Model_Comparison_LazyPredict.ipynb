```python
import pandas as pd
import numpy as np
from charset_normalizer import models
from sklearn.model_selection import train_test_split
from lazypredict.Supervised import LazyClassifier
feature_dataset = pd.read_csv("../data/processed/feature_dataset.csv")
feature_dataset.head()
X = feature_dataset.drop("SepsisLabel", axis=1)
y = feature_dataset["SepsisLabel"]
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42

)
clf = LazyClassifier(
    verbose = 0, #this make the system not to update every little process
    ignore_warnings = True, # if any other model fails it is ignored
    predictions = True

)
models, predictions = clf.fit(
    X_train,
    X_test,
    y_train,
    y_test
) #starts the work.+
print(models)

```
