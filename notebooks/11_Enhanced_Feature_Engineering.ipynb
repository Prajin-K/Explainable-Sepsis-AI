```python
selected_features = list(dict.fromkeys(selected_features))
print(len(selected_features))
print(len(set(selected_features)))
```

```python
model_a = timeseries[selected_features].copy()

print(model_a.shape)
print("Duplicate Columns:", model_a.columns.duplicated().sum())
```

```python
# Shock Index = HR / SBP

model_a["shock_index"] = (
    model_a["heart_rate"] /
    model_a["sbp"]
)

model_a[
    ["heart_rate", "sbp", "shock_index"]
].head(10) # above 1 sepsis risk and between 0.7 to 1 is suspicious of sepsis
```

```python
# mean atrial pressure
model_a["map_calculated"] = (
    model_a["dbp"] +
    (model_a["sbp"] - model_a["dbp"]) / 3
)

model_a[
    ["sbp", "dbp", "mbp", "map_calculated"]
].head() # below 65 is risk of sepsis
```

```python
#  kidney fn ratio denotes kidneys function because during sepsis kidney gets affected mainly
model_a["bun_creatinine_ratio"] = (
    model_a["bun"] /
    model_a["creatinine"]
)

model_a[
    ["bun", "creatinine", "bun_creatinine_ratio"]
].head()
model_a[
    model_a["bun_creatinine_ratio"].notna()
][["bun", "creatinine", "bun_creatinine_ratio"]].head(10)
```

```python
print(model_a["bun_creatinine_ratio"].notna().sum())
model_a[
    model_a["bun_creatinine_ratio"].notna()
][["bun", "creatinine", "bun_creatinine_ratio"]].head(10)
```

```python
model_a["spo2_deficit"] = (
    100 -
    model_a["spo2"]
)

model_a[
    ["spo2", "spo2_deficit"]
].head()
```

```python
print("="*50)
print("Model Shape:", model_a.shape)
print("Number of Features:", model_a.shape[1])

print("\nEngineered Features Added:")
engineered = [
    "shock_index",
    "pulse_pressure",
    "map_calculated",
    "bun_creatinine_ratio",
    "spo2_deficit"
]

for feature in engineered:
    print("✓", feature)

print("="*50)
```

```python
# heart rate change over last hour
model_a = model_a.sort_values(
    ["stay_id","hour"]
).reset_index(drop=True)
```

```python
model_a["heart_rate_prev"] = (
    model_a
    .groupby("stay_id")["heart_rate"] # groupby is used to treat every stay id as seperate one and [heart rate] to choose the category from the table
    .shift(1) # shift (1) used to choose the previous one of the category
)
```

```python
model_a[
    [
        "stay_id",
        "hour",
        "heart_rate",
        "heart_rate_prev"
    ]
].head(15)
```

```python
# heart rate change
model_a["heart_rate_delta"] = (
    model_a["heart_rate"] -
    model_a["heart_rate_prev"]
)
```

```python
model_a[
    [
        "heart_rate",
        "heart_rate_prev",
        "heart_rate_delta" # delta fn is used to identify the trend instead of flase alarming because sudden heart rate rise
    ]
].head(10)
```

```python
# respiratory rate
model_a["resp_rate_prev"] = (
    model_a
    .groupby("stay_id")["resp_rate"]
    .shift(1)
)
```

```python
model_a["resp_rate_delta"] = (
    model_a["resp_rate"] -
    model_a["resp_rate_prev"]
)

model_a[
    [
        "stay_id",
        "hour",
        "resp_rate",
        "resp_rate_prev",
        "resp_rate_delta"
    ]
].head(15)
```

```python
# previous temperature
model_a["temperature_prev"] = (
    model_a
    .groupby("stay_id")["temperature"]
    .shift(1)
)
model_a["temperature_delta"] = (
    model_a["temperature"] -
    model_a["temperature_prev"]
)
model_a[
    [
        "stay_id",
        "hour",
        "temperature",
        "temperature_prev",
        "temperature_delta"
    ]
].head(15)
```

```python
# previous sbp
model_a["sbp_prev"] = (
    model_a
    .groupby("stay_id")["sbp"]
    .shift(1)
)
model_a["sbp_delta"] = (
    model_a["sbp"] -
    model_a["sbp_prev"]
)
model_a[
    [
        "stay_id",
        "hour",
        "sbp",
        "sbp_prev",
        "sbp_delta"
    ]
].head(15)
```

```python
# previous map
model_a["mbp_prev"] = (
    model_a
    .groupby("stay_id")["mbp"]
    .shift(1)
)
model_a["mbp_delta"] = (
    model_a["mbp"] -
    model_a["mbp_prev"]
)
model_a[
    [
        "stay_id",
        "hour",
        "mbp",
        "mbp_prev",
        "mbp_delta"
    ]
].head(15)
```

```python
# previous spo2
model_a["spo2_prev"] = (
    model_a
    .groupby("stay_id")["spo2"]
    .shift(1)
)
model_a["spo2_delta"] = (
    model_a["spo2"] -
    model_a["spo2_prev"]
)
model_a[
    [
        "stay_id",
        "hour",
        "spo2",
        "spo2_prev",
        "spo2_delta"
    ]
].head(15)
```

```python
#glasgow coma scale measure a person's level of consciousness after a brain injury or during critical illness (like sepsis)
model_a["gcs_prev"] = (
    model_a
    .groupby("stay_id")["gcs"]
    .shift(1)
)
model_a["gcs_delta"] = (
    model_a["gcs"] -
    model_a["gcs_prev"]
)
model_a[
    [
        "stay_id",
        "hour",
        "gcs",
        "gcs_prev",
        "gcs_delta"

    ]
].head(15)
```

```python
#ROLLING FEATURE mean
model_a["heart_rate_roll3_mean"] = (
    model_a
    .groupby("stay_id")["heart_rate"]
    .transform(
        lambda x: x.rolling(
            window = 3,
            min_periods = 1,

        ).mean()
    )
)
```

```python
#ROLLING FEATURE mean
model_a["heart_rate_roll6_mean"] = (
    model_a
    .groupby("stay_id")["heart_rate"]
    .transform(
        lambda x: x.rolling(
            window = 6,
            min_periods = 1,

        ).mean()
    )
)
```

```python
model_a[
    [
        "stay_id",
        "hour",
        "heart_rate",
        "heart_rate_roll3_mean"
    ]
].head(15)
```

```python
# ROllING STD
model_a["heart_rate_roll3_std"] = (
    model_a
    .groupby("stay_id")["heart_rate"]
    .transform(
        lambda x: x.rolling(
            window=3,
            min_periods=1
        ).std()
    )
)
```

```python
# ROLLING MAXIMUM
model_a["heart_rate_roll3_max"] = (
    model_a
    .groupby("stay_id")["heart_rate"]
    .transform(
        lambda x: x.rolling(
            window=3,
            min_periods=1
        ).max()
    )
)
```

```python
# ROLLING MINIMUM
model_a["heart_rate_roll3_min"] = (
    model_a
    .groupby("stay_id")["heart_rate"]
    .transform(
        lambda x: x.rolling(
            window=3,
            min_periods=1
        ).min()
    )
)
```

```python
# 3-hour rolling mean
model_a["resp_rate_roll3_mean"] = (
    model_a.groupby("stay_id")["resp_rate"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).mean())
)

# 6-hour rolling mean
model_a["resp_rate_roll6_mean"] = (
    model_a.groupby("stay_id")["resp_rate"]
    .transform(lambda x: x.rolling(window=6, min_periods=1).mean())
)

# 3-hour rolling std
model_a["resp_rate_roll3_std"] = (
    model_a.groupby("stay_id")["resp_rate"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).std())
)

# 3-hour rolling max
model_a["resp_rate_roll3_max"] = (
    model_a.groupby("stay_id")["resp_rate"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).max())
)

# 3-hour rolling min
model_a["resp_rate_roll3_min"] = (
    model_a.groupby("stay_id")["resp_rate"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).min())
)
```

```python
# 3-hour rolling mean
model_a["sbp_roll3_mean"] = (
    model_a.groupby("stay_id")["sbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).mean())
)

# 6-hour rolling mean
model_a["sbp_roll6_mean"] = (
    model_a.groupby("stay_id")["sbp"]
    .transform(lambda x: x.rolling(window=6, min_periods=1).mean())
)

# 3-hour rolling std
model_a["sbp_roll3_std"] = (
    model_a.groupby("stay_id")["sbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).std())
)

# 3-hour rolling max
model_a["sbp_roll3_max"] = (
    model_a.groupby("stay_id")["sbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).max())
)

# 3-hour rolling min
model_a["sbp_roll3_min"] = (
    model_a.groupby("stay_id")["sbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).min())
)
```

```python
# 3-hour rolling mean
model_a["mbp_roll3_mean"] = (
    model_a.groupby("stay_id")["mbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).mean())
)

# 6-hour rolling mean
model_a["mbp_roll6_mean"] = (
    model_a.groupby("stay_id")["mbp"]
    .transform(lambda x: x.rolling(window=6, min_periods=1).mean())
)

# 3-hour rolling std
model_a["mbp_roll3_std"] = (
    model_a.groupby("stay_id")["mbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).std())
)

# 3-hour rolling max
model_a["mbp_roll3_max"] = (
    model_a.groupby("stay_id")["mbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).max())
)

# 3-hour rolling min
model_a["mbp_roll3_min"] = (
    model_a.groupby("stay_id")["mbp"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).min())
)
```

```python
# 3-hour rolling mean
model_a["spo2_roll3_mean"] = (
    model_a.groupby("stay_id")["spo2"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).mean())
)

# 6-hour rolling mean
model_a["spo2_roll6_mean"] = (
    model_a.groupby("stay_id")["spo2"]
    .transform(lambda x: x.rolling(window=6, min_periods=1).mean())
)

# 3-hour rolling std
model_a["spo2_roll3_std"] = (
    model_a.groupby("stay_id")["spo2"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).std())
)

# 3-hour rolling max
model_a["spo2_roll3_max"] = (
    model_a.groupby("stay_id")["spo2"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).max())
)

# 3-hour rolling min
model_a["spo2_roll3_min"] = (
    model_a.groupby("stay_id")["spo2"]
    .transform(lambda x: x.rolling(window=3, min_periods=1).min())
)
```
